AVALIAÇÃO SQLITE

```python
import sqlite3
```

```python
nome = 'escola.db'
```

CRIANDO TABELA SQLITE

```python
script_alunos = """CREATE TABLE IF NOT EXISTS Alunos (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                nome TEXT NOT NULL,
                idade INTEGER NOT NULL,
                email TEXT NOT NULL
            );"""
            

try:
    with sqlite3.connect(nome) as conn:
        cur = conn.cursor()

        cur.execute(script_alunos)

        conn.commit()

        print("Tabelas Criadas com Sucesso")
except sqlite3.OperationalError as e:
    print("ERRO: ", e)
```

INSERINDO REGISTROS

```python
sql = "INSERT INTO Alunos (nome, idade, email) VALUES (?,?,?)"

try:
    with sqlite3.connect(nome) as conn:
            
        cur = conn.cursor()

        cur.execute(sql, ('Nicolly', '16', 'nick@email.com'))

        conn.commit()
        


except sqlite3.OperationalError as e:
    print("ERRO: ", e)
```

LISTANDO REGISTROS

```python
sql = "SELECT * FROM Alunos"


try:
    with sqlite3.connect(nome) as conn:
        
        cur = conn.cursor()
       
        cur.execute(sql)
        res = cur.fetchall() 

        print(res)
except sqlite3.OperationalError as e:
    print("ERRO: ", e)
```

ATUALIZANDO REGISTROS

```python
sql = "UPDATE Alunos SET idade = ? WHERE id = ?"

try:
    with sqlite3.connect(nome) as conn:
            
        cur = conn.cursor()
       
        cur.execute(sql, (20, 3))

        conn.commit()


except sqlite3.OperationalError as e:
    print("ERRO: ", e)
```

DELETANDO REGISTROS

```python
sql = "DELETE FROM Alunos WHERE id = ?"

try:
    with sqlite3.connect(nome) as conn:
            
        cur = conn.cursor()

        cur.execute(sql, (3,))

        conn.commit()
        


except sqlite3.OperationalError as e:
    print("ERRO: ", e)
```
